```python
import sqlite3

# Connect to database
conn = sqlite3.connect("expense_tracker.db")
cursor = conn.cursor()

# Create expenses table
cursor.execute("""
CREATE TABLE IF NOT EXISTS expenses (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
        date TEXT NOT NULL,
            category TEXT NOT NULL,
                description TEXT,
                    amount REAL NOT NULL,
                        payment_method TEXT
                        )
                        """)

conn.commit()

print("Database and expenses table created successfully!")
```

```python
cursor.execute("""
INSERT INTO expenses (date, category, description, amount, payment_method)
VALUES (?, ?, ?, ?, ?)
""", ("2026-09-23", "Food", "Lunch", 150, "UPI"))

conn.commit()

print("Expense added successfully!")
```

```python
expenses = [
      ("2026-09-23", "Travel", "Auto fare", 120, "Cash"),
          ("2026-09-22", "Shopping", "Clothes", 800, "UPI"),
              ("2026-09-21", "Food", "Dinner", 250, "UPI"),
                  ("2026-09-20", "Education", "Notebook", 100, "Cash")
                  ]
cursor.executemany("""
                  INSERT INTO expenses (date, category, description, amount, payment_method)
                  VALUES (?, ?, ?, ?, ?)
                  """, expenses)

conn.commit()

print("Multiple expenses added successfully!")
```

```python
cursor.execute("SELECT * FROM expenses")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
import pandas as pd

cursor.execute("SELECT * FROM expenses")
rows = cursor.fetchall()

columns = ["ID", "Date", "Category", "Description", "Amount", "Payment Method"]

df = pd.DataFrame(rows, columns=columns)

df
```

```python
category = "Food"

cursor.execute("""
SELECT * FROM expenses
WHERE category = ?
""", (category,))

results = cursor.fetchall()

for row in results:
    print(row)
```

```python
cursor.execute("SELECT SUM(amount) FROM expenses")

total = cursor.fetchone()[0]

print("Total Expenses: ₹", total)
```

```python
cursor.execute("""
SELECT category, SUM(amount) AS total_amount
FROM expenses
GROUP BY category
""")

results = cursor.fetchall()

for category, total in results:
    print(category, "→ ₹", total)
```

```python
import sqlite3

conn = sqlite3.connect("expense_tracker.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS expenses (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
        date TEXT NOT NULL,
            category TEXT NOT NULL,
                description TEXT,
                    amount REAL NOT NULL,
                        payment_method TEXT
                        )

                        """)
conn.commit()

print("Database ready!")
```

```python
expenses = [
      ("2026-09-23", "Food", "Lunch", 150, "UPI"),
          ("2026-09-23", "Travel", "Auto fare", 120, "Cash"),
              ("2026-09-22", "Shopping", "Clothes", 800, "UPI"),
                  ("2026-09-21", "Food", "Dinner", 250, "UPI"),
                      ("2026-09-20", "Education", "Notebook", 100, "Cash")
                      ]

cursor.executemany("""
                      INSERT INTO expenses
                      (date, category, description, amount, payment_method)
                      VALUES (?, ?, ?, ?, ?)
                      """, expenses)

conn.commit()

print("5 expenses added successfully!")
```

- INSERT

To develop a simple database-driven application using Python and SQL for managing and analyzing expense information.
